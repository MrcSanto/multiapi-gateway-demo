```python
def calcular_latencia_media(df_latencia):
    return df_latencia['Latency'].mean()

arquivos = {
    'GO': [
        results_GO_1, results_GO_2, results_GO_3
    ],
    'Node': [
        results_NODE_1, results_NODE_2, results_NODE_3
    ],
    'Python': [
        results_Python_1, results_Python_2, results_Python_3
    ]
}


dados = []
for api, lista_arquivos in arquivos.items():
    for i, arquivo in enumerate(lista_arquivos, start=1):
        latencia = calcular_latencia_media(arquivo)
        dados.append({'API': api, 'Rodada': f'Rodada {i}', 'Latência Média (ms)': latencia})


df_latencias = pd.DataFrame(dados)

# Exibir tabela
print(df_latencias)

# Gráfico interativo
fig = px.bar(df_latencias, x='Rodada', y='Latência Média (ms)', color='API', barmode='group',
             title='Latência Média por API e Rodada')
# fig.show(renderer="browser")
```

```python
# Função para calcular métricas do summary
def calcular_metricas_summary(df_summary):
    avg_tempo = df_summary['Average'].mean()
    avg_throughput = df_summary['Throughput'].mean()
    avg_erros = df_summary['Error %'].apply(lambda x: float(x.strip('%'))).mean()
    return avg_tempo, avg_throughput, avg_erros

# Dicionário com os DataFrames
apis = {
    'GO': [(summary_GO_1, results_GO_1), (summary_GO_2, results_GO_2), (summary_GO_3, results_GO_3)],
    'Node': [(summary_NODE_1, results_NODE_1), (summary_NODE_2, results_NODE_2), (summary_NODE_3, results_NODE_3)],
    'Python': [(summary_Python_1, results_Python_1), (summary_Python_2, results_Python_2), (summary_Python_3, results_Python_3)]
}

# Construir lista de dados
dados = []
dados_erros = []

for api, rodadas in apis.items():
    for i, (summary_df, results_df) in enumerate(rodadas, start=1):
        avg_tempo, avg_throughput, avg_erros = calcular_metricas_summary(summary_df)
        latencia_media = calcular_latencia_media(results_df)
        
        dados.append({
            'API': api,
            'Rodada': i,
            'Tempo Médio (ms)': round(avg_tempo, 2),
            'Throughput (req/s)': round(avg_throughput, 2),
            'Erro (%)': round(avg_erros, 4),
            'Latência Média (ms)': round(latencia_media, 2)
        })
        
        # Coletar dados de erros detalhados
        for _, row in summary_df.iterrows():
            erro_percentual = float(row['Error %'].strip('%'))
            total_samples = row['# Samples']
            total_erros = int((erro_percentual / 100) * total_samples)
            
            dados_erros.append({
                'API': api,
                'Rodada': i,
                'Label': row['Label'],
                'Total Requisições': total_samples,
                'Total Erros': total_erros,
                'Erro (%)': round(erro_percentual, 4),
                'Throughput': row['Throughput']
            })

# Criar DataFrame consolidado
df_resumo = pd.DataFrame(dados)
df_erros_detalhado = pd.DataFrame(dados_erros)

# 1. Resumo de erros por API (todas as rodadas)
df_erros_por_api = df_erros_detalhado.groupby('API').agg({
    'Total Requisições': 'sum',
    'Total Erros': 'sum',
}).reset_index()
df_erros_por_api['Taxa de Erro (%)'] = (
    df_erros_por_api['Total Erros'] / df_erros_por_api['Total Requisições'] * 100
).round(4)

print("\n" + "="*60)
print("RESUMO DE ERROS POR API")
print("="*60)
print(df_erros_por_api.to_string(index=False))

# 2. Erros por API e Rodada
df_erros_por_rodada = df_erros_detalhado.groupby(['API', 'Rodada']).agg({
    'Total Requisições': 'sum',
    'Total Erros': 'sum',
}).reset_index()
df_erros_por_rodada['Taxa de Erro (%)'] = (
    df_erros_por_rodada['Total Erros'] / df_erros_por_rodada['Total Requisições'] * 100
).round(4)

print("\n" + "="*60)
print("ERROS POR API E RODADA")
print("="*60)
print(df_erros_por_rodada.to_string(index=False))

# 3. Endpoints com mais erros
df_endpoints_com_erros = df_erros_detalhado[df_erros_detalhado['Total Erros'] > 0].copy()
df_endpoints_com_erros = df_endpoints_com_erros.sort_values('Total Erros', ascending=False)

print("\n" + "="*60)
print("ENDPOINTS COM MAIS ERROS")
print("="*60)
print(df_endpoints_com_erros[['API', 'Rodada', 'Label', 'Total Erros', 'Erro (%)']].to_string(index=False))


# Gráfico 1: Tempo Médio por API e Rodada
fig_tempo = px.bar(df_resumo, x='Rodada', y='Tempo Médio (ms)', color='API', barmode='group',
                   title='Tempo Médio por API e Rodada')
fig_tempo.show()

# Gráfico 2: Latência Média por API e Rodada
fig_latencia = px.bar(df_resumo, x='Rodada', y='Latência Média (ms)', color='API', barmode='group',
                      title='Latência Média por API e Rodada')
fig_latencia.show()

# Gráfico 3: Throughput por API e Rodada
fig_throughput = px.bar(df_resumo, x='Rodada', y='Throughput (req/s)', color='API', barmode='group',
                        title='Throughput por API e Rodada')
fig_throughput.show()

# Gráfico 4: Taxa de Erro por API
fig_erro_api = px.bar(df_erros_por_api, x='API', y='Taxa de Erro (%)',
                      title='Taxa de Erro por API (Todas as Rodadas)',
                      text='Taxa de Erro (%)')
fig_erro_api.update_traces(texttemplate='%{text:.4f}%', textposition='outside')
fig_erro_api.show()

# Gráfico 5: Taxa de Erro por API e Rodada
fig_erro_rodada = px.bar(df_erros_por_rodada, x='Rodada', y='Taxa de Erro (%)', 
                         color='API', barmode='group',
                         title='Taxa de Erro por API e Rodada')
fig_erro_rodada.show()

# Gráfico 6: Total de Erros por API e Rodada
fig_total_erros = px.bar(df_erros_por_rodada, x='Rodada', y='Total Erros', 
                         color='API', barmode='group',
                         title='Total de Erros por API e Rodada')
fig_total_erros.show()

df_erros_por_api.to_csv('relatorio_erros_por_api.csv', index=False)
```
